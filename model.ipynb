```python
# Map each file to the specific columns you want from it (include the join key!)
files_and_cols = {
    "data/DIQ_L.csv": ["SEQN", "DIQ010", "DID040", "DIQ160"],
    
    "data/DBQ_L.csv": ["SEQN", "DBQ301", "DBQ330", "DBQ360", "DBD381", "DBD411", "DBQ424", "DBQ930", "DBQ940"],
    
    "data/FSQ_L.csv": ["SEQN", "FSD032A", "FSD032B", "FSD032C", "FSD041", "FSD052", "FSD061", "FSD071", "FSD081", "FSD092", "FSD151", "FSQ165", "FSQ012", "FSD162"],
    
    "data/ALQ_L.csv": ["SEQN", "ALQ111", "ALQ121", "ALQ130"],
    
    "data/BPQ_L.csv": ["SEQN", "BPQ020", "BPQ030", "BPQ080"],
    
    "data/INQ_L.csv": ["SEQN", "INDFMMPI", "INDFMMPC"],
    
    "data/OCQ_L.csv": ["SEQN", "OCD150", "OCQ180", "OCQ210", "OCQ215"],
    
    "data/PAQ_L.csv": ["SEQN", "PAD790Q", "PAD790U", "PAD800", "PAD810Q", "PAD810U", "PAD820", "PAD680"],
    
    "data/WHQ_L.csv": ["SEQN", "WHD010", "WHD020", "WHD050", "WHQ070"]
}

# Scan each file lazily
lazy_frames = [
    pl.scan_csv(file).select(cols) for file, cols in files_and_cols.items()
]

# Start with the first file and left-join the rest on "SEQN"
combined_lazy = lazy_frames[0]
for lf in lazy_frames[1:]:
  combined_lazy = combined_lazy.join(lf, on="SEQN", how="left")

# Execute the entire pipeline in parallel
df = combined_lazy.collect()
```
